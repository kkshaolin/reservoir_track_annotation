## การนำมาใช้งานกับ YOLO

````python
import json
import numpy as np

class DatasetToYOLOAdapter:
    """
    Adapter Class สำหรับแปลง Label จาก Cityscapes, CULane, และ TuSimple
    ให้อยู่ในฟอร์แมต YOLOv8-seg / YOLOv11-seg: <class_id> <x1> <y1> <x2> <y2> ... (Normalized 0-1)
    """
    def __init__(self, img_width=1280, img_height=720):
        self.w = img_width
        self.h = img_height

    def cityscapes_polygon_to_yolo(self, polygon_coords, class_id=0):
        """แปลงพิกัด Polygon [[x1, y1], [x2, y2], ...] เป็น YOLO Normalized String"""
        normalized = []
        for x, y in polygon_coords:
            norm_x = max(0.0, min(1.0, x / self.w))
            norm_y = max(0.0, min(1.0, y / self.h))
            normalized.extend([f"{norm_x:.6f}", f"{norm_y:.6f}"])
        return f"{class_id} " + " ".join(normalized)

    def tusimple_lanes_to_yolo(self, lanes, h_samples, class_id=1, thickness=10):
        """แปลง TuSimple JSON (จุดพิกัดเส้นเลน) เป็น Polygon ล้อมรอบเส้นตามมาตรฐาน YOLO-seg"""
        yolo_labels = []
        for lane in lanes:
            points = [[x, y] for x, y in zip(lane, h_samples) if x > 0]
            if len(points) < 2:
                continue
            
            # ขยายเส้น (Polyline) ออกซ้าย-ขวาเพื่อสร้าง Mask Area ความหนาเท่ากับ thickness
            left_edge = [[x - thickness / 2, y] for x, y in points]
            right_edge = [[x + thickness / 2, y] for x, y in reversed(points)]
            poly = left_edge + right_edge
            
            yolo_labels.append(self.cityscapes_polygon_to_yolo(poly, class_id=class_id))
        return yolo_labels

# ตัวอย่างการทดสอบฟังก์ชัน Adapter
adapter = DatasetToYOLOAdapter(img_width=1280, img_height=720)
dummy_polygon = [[100, 200], [150, 400], [250, 400], [200, 200]]
print("ตัวอย่างผลลัพธ์ YOLO-seg Format:")
print(adapter.cityscapes_polygon_to_yolo(dummy_polygon, class_id=0))
````

## Sampling Dataset

ทำการสุ่มภาพในวิดีโอ จำนวน 1000 ภาพจากในวิดีโอ

````python
import cv2
import os
import math
````

````python
def extract_frames_for_yolo(video_path, output_dir, target_frames):
    """
    ฟังก์ชันสำหรับสุ่มภาพจากวิดีโอเพื่อทำ Dataset YOLO
    """
    # สร้างโฟลเดอร์สำหรับเซฟรูปถ้ายังไม่มี
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # เปิดไฟล์วิดีโอ
    cap = cv2.VideoCapture(video_path)
    
    if not cap.isOpened():
        print(f"ไม่สามารถเปิดไฟล์ได้")
        return

    # ดึงข้อมูลพื้นฐานของวิดีโอ
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    # คำนวณว่าจะต้องข้ามทีละกี่เฟรม
    step = max(1, math.floor(total_frames / target_frames))

    saved_count = 0
    current_frame = 0

    while True:
        ret, frame = cap.read()
        
        # ถ้าอ่านวิดีโอจบหรือมีภาพครบเฟรมแล้ว ให้ออกจากลูป
        if not ret or saved_count >= target_frames:
            break

        # ตรวจสอบว่าเฟรมปัจจุบันถึงรอบที่จะบันทึกหรือยัง
        if current_frame % step == 0:
            # ตั้งชื่อไฟล์เป็นตัวเลขที่เรียงกัน เช่น 0001.jpg, 0002.jpg (YOOLO ชอบตัวเลขเรียง)
            filename = f"frame_{saved_count:05d}.jpg"
            filepath = os.path.join(output_dir, filename)
            
            # บันทึกภาพ
            cv2.imwrite(filepath, frame)
            saved_count += 1
            
            # แสดง Progress (แสดงทุกๆ 100 ภาพ)
            if saved_count % 100 == 0:
                print(f"บันทึกแล้ว {saved_count}/{target_frames} ภาพ...")

        current_frame += 1

    # ปิดไฟล์วิดีโอ
    cap.release()
    print(f"บันทึกไฟล์ไปที่ {output_dir} ทั้งหมด {saved_count} ภาพ")
````

````python
# พาธของไฟล์วิดีโอ
VIDEO_FILE = "psu-reservoir\\psu-reservoir-2026Aug06_121427.mp4"
    
# โฟลเดอร์ที่จะเก็บรูปภาพ
OUTPUT_FOLDER = "psu-reservoir\\images\\train" 
    
# จำนวนเฟรมที่ต้องการ
NUM_FRAMES = 1001      

extract_frames_for_yolo(VIDEO_FILE, OUTPUT_FOLDER, NUM_FRAMES)
````

````python
# แปลง Annotation XML จาก CVAT เป็นไฟล์ YOLO Segmentation .txt
import xml.etree.ElementTree as ET
from shapely.geometry import LineString
import os

def parse_points(points_str):
    return [tuple(map(float, p.split(","))) for p in points_str.strip().split(";") if p]

def polyline_to_polygon(coords, buffer_dist=10.0):
    line = LineString(coords)
    poly = line.buffer(distance=buffer_dist, cap_style="flat", join_style="round")
    if poly.geom_type == "MultiPolygon":
        poly = max(poly.geoms, key=lambda a: a.area)
    return list(poly.exterior.coords)

print("สคริปต์ Polyline to Polygon พร้อมใช้งานสำหรับการแปลง Label")
````

````python
import glob
import os
import random
import cv2
import matplotlib.pyplot as plt

# 1. กำหนด Path โฟลเดอร์ภาพและ Label
IMAGE_DIR = "c:\\Users\\kitti\\Desktop\\bbox\\data\\images\\train\\"
LABEL_DIR = "c:\\Users\\kitti\\Desktop\\bbox\\data\\labels\\train\\"

CLASSES = ["class_0", "class_1","class_2", "class_3","class_4"]


def draw_yolo_bboxes(img_path, label_path):
    img = cv2.imread(img_path)
    if img is None:
        print(f"Error: ไม่สามารถโหลดรูป {img_path} ได้")
        return

    # แปลงสี BGR เป็น RGB สำหรับ Matplotlib
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w, _ = img.shape

    if not os.path.exists(label_path):
        print(f"Warning: ไม่พบไฟล์ label ที่ {label_path}")
        return

    with open(label_path, "r") as f:
        lines = f.readlines()

    for line in lines:
        parts = line.strip().split()
        if len(parts) < 5:
            continue

        class_id = int(parts[0])
        # พิกัด Normalize (0 - 1)
        x_c, y_c, bw, bh = map(float, parts[1:5])

        # แปลงเป็นพิกัดพิกเซลจริง (xmin, ymin, xmax, ymax)
        xmin = int((x_c - bw / 2) * w)
        ymin = int((y_c - bh / 2) * h)
        xmax = int((x_c + bw / 2) * w)
        ymax = int((y_c + bh / 2) * h)

        # วาด Bounding Box
        cv2.rectangle(img, (xmin, ymin), (xmax, ymax), (0, 255, 0), 2)

        # ใส่ข้อความชื่อคลาส
        label_text = (
            CLASSES[class_id] if class_id < len(CLASSES) else str(class_id)
        )
        cv2.putText(
            img,
            label_text,
            (xmin, max(ymin - 8, 15)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (255, 0, 0),
            2,
        )

    plt.figure(figsize=(10, 6))
    plt.imshow(img)
    plt.title(f"Review: {os.path.basename(img_path)}")
    plt.axis("off")
    plt.show()


# สุ่มตรวจ 1 รูป
image_list = glob.glob(os.path.join(IMAGE_DIR, "*.*"))
if image_list:
    sample_img = random.choice(image_list)
    base_name = os.path.splitext(os.path.basename(sample_img))[0]
    sample_label = os.path.join(LABEL_DIR, f"{base_name}.txt")

    draw_yolo_bboxes(sample_img, sample_label)
else:
    print(f"ไม่พบไฟล์รูปภาพในโฟลเดอร์ {IMAGE_DIR}")
````

````python
import glob
import os
import random
import cv2
import matplotlib.pyplot as plt
import numpy as np

IMAGE_DIR = "psu-reservoir\\SamplingDataset"
LABEL_DIR = "labels_yolo"

CLASSES = ["class_0", "class_1","class_2", "class_3","class_4"]


def draw_yolo_annotations(img_path, label_path):
    img = cv2.imread(img_path)
    if img is None:
        print(f"Error: ไม่สามารถเปิดรูป {img_path} ได้")
        return

    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w, _ = img.shape
    overlay = img.copy()

    if not os.path.exists(label_path):
        print(f"Warning: ไม่พบไฟล์ label ที่ {label_path}")
        return

    with open(label_path, "r") as f:
        lines = f.readlines()

    for line in lines:
        parts = line.strip().split()
        if len(parts) < 5:
            continue

        class_id = int(parts[0])
        coords = list(map(float, parts[1:]))
        label_text = (
            CLASSES[class_id] if class_id < len(CLASSES) else str(class_id)
        )

        # ตรวจสอบรูปแบบ: 4 พิกัด = Bounding Box, > 4 พิกัด = Segmentation Polygon
        if len(coords) == 4:
            x_c, y_c, bw, bh = coords
            xmin = int((x_c - bw / 2) * w)
            ymin = int((y_c - bh / 2) * h)
            xmax = int((x_c + bw / 2) * w)
            ymax = int((y_c + bh / 2) * h)

            cv2.rectangle(img, (xmin, ymin), (xmax, ymax), (0, 255, 0), 2)
            cv2.putText(
                img,
                label_text,
                (xmin, max(ymin - 8, 15)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (255, 0, 0),
                2,
            )

        elif len(coords) >= 6 and len(coords) % 2 == 0:
            # แปลง x, y normalized เป็นพิกเซลจริง
            pts = []
            for i in range(0, len(coords), 2):
                px = int(coords[i] * w)
                py = int(coords[i + 1] * h)
                pts.append([px, py])

            pts = np.array(pts, dtype=np.int32).reshape((-1, 1, 2))

            # วาดเส้นขอบ Polygon
            cv2.polylines(img, [pts], isClosed=True, color=(0, 255, 0), thickness=2)
            # เติมสีโปร่งแสงลงในพื้นที่ Mask
            cv2.fillPoly(overlay, [pts], color=(0, 255, 0))

            # วาง Label เหนือจุดแรกของ Polygon
            first_pt = pts[0][0]
            cv2.putText(
                img,
                label_text,
                (first_pt[0], max(first_pt[1] - 8, 15)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (255, 0, 0),
                2,
            )

    # ผสมเลเยอร์โปร่งแสง (Alpha Blending)
    img = cv2.addWeighted(overlay, 0.3, img, 0.7, 0)

    plt.figure(figsize=(10, 6))
    plt.imshow(img)
    plt.title(f"Review Annotation: {os.path.basename(img_path)}")
    plt.axis("off")
    plt.show()


# ทดสอบสุ่มเปิดดู 1 รูป
image_list = glob.glob(os.path.join(IMAGE_DIR, "*.*"))
if image_list:
    sample_img = random.choice(image_list)
    base_name = os.path.splitext(os.path.basename(sample_img))[0]
    sample_label = os.path.join(LABEL_DIR, f"{base_name}.txt")

    draw_yolo_annotations(sample_img, sample_label)
else:
    print(f"ไม่พบไฟล์รูปภาพใน {IMAGE_DIR}")
````
