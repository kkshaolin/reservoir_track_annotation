## การนำมาใช้งานกับ YOLO

### หลักการทำงานและการแปลงข้อมูลเข้าสู่ YOLOv8-seg / YOLOv11-seg

โมเดลในกลุ่ม **YOLO Segmentation (YOLOv8-seg และ YOLOv11-seg)** ของ Ultralytics กำหนดให้ไฟล์ Label ต้องเป็นไฟล์ข้อความ `.txt` โดย 1 แถวแทน 1 วัตถุ และพิกัดต้องเป็นรูปหลายเหลี่ยมปิด (Polygon) ที่ถูกแปลงค่าให้อยู่ในช่วง **Normalized Coordinates (0.0 ถึง 1.0)** ทั้งหมด ดังนี้:

$$\text{<class\_id>} \quad x_1 \quad y_1 \quad x_2 \quad y_2 \quad \dots \quad x_n \quad y_n$$

คลาส `DatasetToYOLOAdapter` ถูกสร้างขึ้นเพื่อทำหน้าที่เป็น Adapter แปลงโครงสร้างข้อมูลเดิมของชุดข้อมูลต่างๆ เข้าสู่ฟอร์แมตดังกล่าว:

---

#### 1. การแปลงพิกัด Polygon จาก Cityscapes (`cityscapes_polygon_to_yolo`)
* **ลักษณะข้อมูลเดิม:** จัดเก็บพิกัดขอบเขตวัตถุในรูปแบบชุดพิกเซลจริง (Absolute Pixel Coordinates) เช่น `[[x1, y1], [x2, y2], ...]`
* **ขั้นตอนการแปลง:**
  * นำพิกัดแกน $x$ หารด้วยความกว้างของภาพ ($W = 1280$) จะได้ $x_{norm} = \frac{x}{W}$
  * นำพิกัดแกน $y$ หารด้วยความสูงของภาพ ($H = 720$) จะได้ $y_{norm} = \frac{y}{H}$
  * ป้องกันพิกัดหลุดขอบภาพด้วย `min/max` ให้อยู่ระหว่าง $0.0 - 1.0$ และบันทึกทศนิยม 6 ตำแหน่ง ต่อท้าย `class_id`

---

#### 2. การแปลงเส้นเลนจาก TuSimple (`tusimple_lanes_to_yolo`)
* **ลักษณะข้อมูลเดิม:** TuSimple ไม่ได้เก็บเป็นพื้นที่ (Polygon) แต่เก็บเป็นจุดบนแนวเส้นตรง (Polyline) ผ่านคีย์ `lanes` และ `h_samples` (หากจุดใดไม่มีเส้นลากผ่านจะแทนด้วยค่า `-2`)
* **ขั้นตอนการแปลง:**
  1. **คัดกรองพิกัด:** กรองเฉพาะคู่พิกัด $(x, y)$ ที่มีค่า $x > 0$ เพื่อตัดจุด `-2` ทิ้ง
  2. **ขยายเส้นให้มีความหนา (Polyline Buffer):** เนื่องจากโมเดล Segmentation ต้องการพื้นที่รูปปิด จึงขยายพิกัดออกด้านข้างซ้าย $(x - \frac{\text{thickness}}{2})$ และขวา $(x + \frac{\text{thickness}}{2})$ ตามความหนาของเส้นที่กำหนด (`thickness`)
  3. **เชื่อมจุดเป็น Polygon ปิด:** นำจุดขอบซ้ายมาเรียงต่อกับจุดขอบขวาย้อนกลับ (`reversed(points)`) เพื่อให้ได้โครงร่างหลายเหลี่ยมที่ล้อมรอบแนวเส้นเดิม
  4. **Normalize:** ส่งพิกัดรูปปิดเข้าฟังก์ชัน `cityscapes_polygon_to_yolo` เพื่อแปลงเป็นค่าสเกล $0-1$

---

#### 3. แนวทางสำหรับ CULane (เพิ่มเติม)
* ชุดข้อมูล CULane จัดเก็บแนวเส้นด้วยพิกัด Spline ในไฟล์ `.txt` สามารถประยุกต์ใช้วิธีเดียวกันกับ TuSimple โดยการสร้างแนวเส้นขอบคู่ขนาน (Line Buffer) เพื่อเพิ่มความหนาให้เส้นกลายเป็นรูปหลายเหลี่ยมปิด แล้วคำนวณ Normalize พิกัดเข้าสู่สเกล $0-1$ ก่อนส่งเข้าเทรนใน YOLO

````python
import json
import numpy as np

class DatasetToYOLOAdapter:
    """
    Adapter Class สำหรับแปลง Label จาก Cityscapes, CULane, และ TuSimple
    ให้อยู่ในฟอร์แมต YOLOv8-seg / YOLOv11-seg: <class_id> <x1> <y1> <x2> <y2> ... (Normalized 0-1)
    """
    def __init__(self, img_width=1280, img_height=720):
        self.w = img_width
        self.h = img_height

    def cityscapes_polygon_to_yolo(self, polygon_coords, class_id=0):
        """แปลงพิกัด Polygon [[x1, y1], [x2, y2], ...] เป็น YOLO Normalized String"""
        normalized = []
        for x, y in polygon_coords:
            norm_x = max(0.0, min(1.0, x / self.w))
            norm_y = max(0.0, min(1.0, y / self.h))
            normalized.extend([f"{norm_x:.6f}", f"{norm_y:.6f}"])
        return f"{class_id} " + " ".join(normalized)

    def tusimple_lanes_to_yolo(self, lanes, h_samples, class_id=1, thickness=10):
        """แปลง TuSimple JSON (จุดพิกัดเส้นเลน) เป็น Polygon ล้อมรอบเส้นตามมาตรฐาน YOLO-seg"""
        yolo_labels = []
        for lane in lanes:
            points = [[x, y] for x, y in zip(lane, h_samples) if x > 0]
            if len(points) < 2:
                continue
            
            # ขยายเส้น (Polyline) ออกซ้าย-ขวาเพื่อสร้าง Mask Area ความหนาเท่ากับ thickness
            left_edge = [[x - thickness / 2, y] for x, y in points]
            right_edge = [[x + thickness / 2, y] for x, y in reversed(points)]
            poly = left_edge + right_edge
            
            yolo_labels.append(self.cityscapes_polygon_to_yolo(poly, class_id=class_id))
        return yolo_labels

# ตัวอย่างการทดสอบฟังก์ชัน Adapter
adapter = DatasetToYOLOAdapter(img_width=1280, img_height=720)
dummy_polygon = [[100, 200], [150, 400], [250, 400], [200, 200]]
print("ตัวอย่างผลลัพธ์ YOLO-seg Format:")
print(adapter.cityscapes_polygon_to_yolo(dummy_polygon, class_id=0))
````

## Sampling Dataset

ทำการสุ่มภาพในวิดีโอ จำนวน 1000 ภาพจากในวิดีโอ

````python
import cv2
import os
import math
````

````python
def extract_frames_for_yolo(video_path, output_dir, target_frames):
    """
    ฟังก์ชันสำหรับสุ่มภาพจากวิดีโอเพื่อทำ Dataset YOLO
    """
    # สร้างโฟลเดอร์สำหรับเซฟรูปถ้ายังไม่มี
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # เปิดไฟล์วิดีโอ
    cap = cv2.VideoCapture(video_path)
    
    if not cap.isOpened():
        print(f"ไม่สามารถเปิดไฟล์ได้")
        return

    # ดึงข้อมูลพื้นฐานของวิดีโอ
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    # คำนวณว่าจะต้องข้ามทีละกี่เฟรม
    step = max(1, math.floor(total_frames / target_frames))

    saved_count = 0
    current_frame = 0

    while True:
        ret, frame = cap.read()
        
        # ถ้าอ่านวิดีโอจบหรือมีภาพครบเฟรมแล้ว ให้ออกจากลูป
        if not ret or saved_count >= target_frames:
            break

        # ตรวจสอบว่าเฟรมปัจจุบันถึงรอบที่จะบันทึกหรือยัง
        if current_frame % step == 0:
            # ตั้งชื่อไฟล์เป็นตัวเลขที่เรียงกัน เช่น 0001.jpg, 0002.jpg (YOOLO ชอบตัวเลขเรียง)
            filename = f"frame_{saved_count:05d}.jpg"
            filepath = os.path.join(output_dir, filename)
            
            # บันทึกภาพ
            cv2.imwrite(filepath, frame)
            saved_count += 1
            
            # แสดง Progress (แสดงทุกๆ 100 ภาพ)
            if saved_count % 100 == 0:
                print(f"บันทึกแล้ว {saved_count}/{target_frames} ภาพ...")

        current_frame += 1

    # ปิดไฟล์วิดีโอ
    cap.release()
    print(f"บันทึกไฟล์ไปที่ {output_dir} ทั้งหมด {saved_count} ภาพ")
````

````python
# พาธของไฟล์วิดีโอ
VIDEO_FILE = "psu-reservoir\\psu-reservoir-2026Aug06_121427.mp4"
    
# โฟลเดอร์ที่จะเก็บรูปภาพ
OUTPUT_FOLDER = "psu-reservoir\\images\\train" 
    
# จำนวนเฟรมที่ต้องการ
NUM_FRAMES = 1001      

extract_frames_for_yolo(VIDEO_FILE, OUTPUT_FOLDER, NUM_FRAMES)
````

   - การทำ Bounding Box จะวาด Bounding Box คลุมปริเวณ lane ,track_line_l ,track_line_c และ track_line_r แล้วใช้ TransT ในการช่วย Track ในทุก frame แล้วตรวจสอบในแต่ละ frame ปรับให้ชิดขอบถนนหรือขอบทางมากที่สุด
   - การทำ Polygon  จะวาด Polygon คลุมปริเวณ lane , sideway ,track_line_l ,track_line_c และ track_line_r แล้วใช้ SAM2 ในการช่วยวาดให้ในแต่ละ frame แล้วตรวจสอบในแต่ละ frame แล้วปรับให้ชิดขอบถนน ขอบวัตถุ หรือขอบทางมากที่สุด
   - การทำ Polyline จะวาด Polyline ของ track_line_l ,track_line_c และ track_line_r แล้ว Track ในทุก frame แล้วตรวจสอบ และปรับให้อยู่กลางเส้นมากที่สุด
